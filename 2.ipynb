2. Create a list with elements of different types.
In a loop: count the number of string elements in the list and output the result.
You can use the type() function or any other option you invent.

```python
list = [3, 2.5, "Hi", 7 > 3, 6 < 1, "Saint-Petersburg", "python"]
n = 0

for i in range(len(list)):
    if type(list[i]) == str:
        n += 1
        
print(n)
```
